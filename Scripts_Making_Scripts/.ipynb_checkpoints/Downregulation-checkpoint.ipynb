```python
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
import pandas as pd
import numpy as np
import copy
import seaborn as sns
from scipy.stats import mannwhitneyu as mwu
from scipy.stats import ttest_ind
from scipy.stats import ttest_rel
from statsmodels.stats.multitest import fdrcorrection
from scipy.stats import wilcoxon
from scipy.optimize import curve_fit
from scipy.stats import fisher_exact
import os
from scipy.stats import combine_pvalues
from scipy.stats import spearmanr,pearsonr
from collections import Counter
import sys
from math import floor
from scipy.stats import norm
from scipy.stats import binomtest

sns.set_style("white")
sns.set(font_scale = 1.5)
sns.set_style("white")

hfont = {'fontname':'Arial'}
plt.rcParams["font.family"] = "Arial"

#Code borrowed heavily from here: https://stackoverflow.com/questions/62375034/find-non-overlapping-area-between-two-kde-plots
plt.rcParams.update(
    {"text.usetex": False}
)

palette = {"Human accelerated":"#E31A1C", "Chimp accelerated":"#0058FF", "Not significant":"grey", "Human":"#E31A1C", "Chimp":"#0058FF"}

#For ML, seems like SpecSup250, Per90, and SpecSup250 + Per90 are the way to go
#If we decide to do the whole controlling for effect size when assigning to genes/species for NC PhyloP, we will need to alter code to do it for just species for ML

gobp = pd.read_csv("GOBP_AccelEvol_Input_FiltForAccelEvol.txt", sep= "\t")
d_BP = {}

for index, row in gobp.iterrows():
    d_BP[row["Term"]] = row["Genes"].split(";")

hpo = pd.read_csv("HPO_AccelEvol_Input_FiltForAccelEvol.txt", sep= "\t")
d_HPO = {}

for index, row in hpo.iterrows():
    d_HPO[row["Term"]] = row["Genes"].split(";")

np.random.seed(10)
x = []
for key in d_BP.keys():
    if len(d_BP[key]) > 20 and len(d_BP[key]) < 200:
        x = d_BP[key] + x
x = list(set(x))

def control(ads, adns, seed = 6):

    # choose bins (quantiles work very well)
    bins = np.quantile(
        ads["Size"],
        q=np.linspace(0, 1, 21)  # 20 bins
    )
    
    
    # drop duplicate bin edges just in case
    bins = np.unique(bins)
    
    # assign bins
    ads["bin"] = pd.cut(ads["Size"], bins=bins, include_lowest=True)
    adns["bin"] = pd.cut(adns["Size"], bins=bins, include_lowest=True)
    
    # how many to sample per bin (match vxcu)
    target_counts = ads["bin"].value_counts()
    
    # downsample vxhu
    adns_matched = (
        adns.copy()
        .groupby("bin", group_keys=False)
        .apply(
            lambda x: x.sample(
                n=min(len(x), target_counts.get(x.name, 0)),
                random_state=seed
            )
        )
        .drop(columns="bin")
    ).copy()
    
    target_props = ads["bin"].value_counts(normalize=True)
    N = len(adns)
    
    def sample_bin(x):
        bin_name = x.name
        if bin_name not in target_props:
            return None
    
        n_bin = int(round(target_props[bin_name] * N))
        if n_bin <= 0:
            return None
    
        # sample *with replacement* only if needed
        replace = n_bin > len(x)
        return x.sample(n=n_bin, replace=replace, random_state=seed)
    
    adns_matched = (
        adns.copy()
        .groupby("bin", group_keys=False)
        .apply(sample_bin)
        .drop(columns="bin")
    ).copy()
    return adns_matched.copy()

import warnings

# Suppress all warnings from Pandas
warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.simplefilter(action='ignore', category=pd.errors.PerformanceWarning)
warnings.simplefilter(action='ignore', category=pd.errors.SettingWithCopyWarning)
warnings.filterwarnings("ignore", category=DeprecationWarning)
```

```python
def do_sign_cons(v, lfc_col = "L2FC Chpreffed", p_col = "padj_mine Chpreffed", fdr_cut = 0.05, z_cut = 1.1, d = d_BP):
    v = v[v[p_col] < fdr_cut]
    v = v[v["gene"].isin(x)]
    print(v.shape)
    upt = v[v["Z-score"] > z_cut].shape[0]
    downt = v[v["Z-score"] < -z_cut].shape[0]
    bp = upt/(upt + downt)
    print(bp, downt, upt)
    bl = downt/upt
    out = []
    for key in d.keys():
        if len(d[key]) > 20 and len(d[key]) < 200:
            vk = v[v["gene"].isin(d[key])]
            if len(vk.index) > 15:
                upk = vk[vk["Z-score"] > z_cut]
                downk = vk[vk["Z-score"] < -z_cut]
                up = upk.shape[0]
                down = downk.shape[0]
                out.append([key, up, down, binomtest(up, up + down, p = bp).pvalue, ";".join(list(upk["gene"]) + list(downk["gene"]))])

    df = pd.DataFrame(out)
    df = df[df[1] + df[2] > 15]
    df["l2fc"] = np.log2((df[1]/df[2])*bl)
    df.columns = ["Term", "Up", "Down", "p-value", "Genes", "l2fc"]
    return df

a = pd.read_csv("AllTissues_MinReads_Per_Allele10Variance_Mean.txt", sep = "\t").dropna()
a = a[a["Size"] > 100]
a = a[a["Variance of ASE Dist"] < 0.5].set_index("gene")[["Variance of ASE Dist", "Median of ASE Dist"]]
v = pd.read_csv("../Myriad_RNA/CM_Filtered.txt", sep = "\t")
v.index = v["gene"]
v = v.join(a).dropna()
v["Median of ASE Dist"] = np.log2(v["Median of ASE Dist"])
v["Z-score"] = (v["L2FC Chpreffed"] - v["Median of ASE Dist"])/np.sqrt(v["Variance of ASE Dist"])
df = do_sign_cons(v)
```

```python
def do_unsign(v, lfc_col = "L2FC Chpreffed", p_col = "padj_mine Chpreffed", fdr_cut = 0.05, lfc_cut = 0.5, d = d_BP):
    v = v[v["gene"].isin(x)]
    tot_meet_cut = v[(v[p_col] < fdr_cut) & (np.abs(v[lfc_col]) > lfc_cut)].shape[0]
    print(tot_meet_cut)
    tot_nmeet_cut = v.shape[0] - tot_meet_cut
    out = []
    for key in d.keys():
        if len(d[key]) > 20 and len(d[key]) < 200:
            vk = v[v["gene"].isin(d_BP[key])]
            if vk.shape[0] > 30:
                meet_cut = vk[(vk[p_col] < fdr_cut) & (np.abs(vk[lfc_col]) > lfc_cut)].shape[0]
                nmeet_cut = vk.shape[0] - meet_cut
                out.append([key, meet_cut, nmeet_cut, fisher_exact([[meet_cut, nmeet_cut], [tot_meet_cut - meet_cut, tot_nmeet_cut - nmeet_cut]], alternative = "greater")[1], fisher_exact([[meet_cut, nmeet_cut], [tot_meet_cut - meet_cut, tot_nmeet_cut - nmeet_cut]], alternative = "greater")[0], ";".join(list(vk[(vk[p_col] < fdr_cut) & (np.abs(vk[lfc_col]) > lfc_cut)]["gene"]))])
    df = pd.DataFrame(out)
    df.columns = ["Term", "Sig", "Nsig", "p-value", "OR", "Genes"]
    return df.sort_values("OR", ascending = False)
```

```python
folder = "../Allen_MTG_Single_Cell/Subclass/Processed"
#folder = "../Ma_Sestan_Single_Cell/Processed"
skel = "DESeq2_REPLACE_Human_Chimp.txt"
output = []
ind = []
if "Allen" in folder:
    celltypes = ['Micro-PVM', 'L2/3 IT', 'Sst Chodl', 'Sst', 'OPC', 'L6 CT', 'Lamp5_Lhx6', 'Chandelier', 'Pvalb', 'L4 IT', 'VLMC', 'Pax6', 'Endo', 'Vip', 'Astro', 'L5 IT', 'L5/6 NP', 'Lamp5', 'Sncg', 'L5 ET', 'Oligo', 'L6 IT Car3', 'L6 IT', 'L6b']
    celltypes = [x.replace(" ", "_").replace("/", "-") for x in celltypes]
else:
    celltypes = ['PVALB', 'SST_NPY', 'Astro', 'L5-6_NP', 'RB', 'VIP', 'Endo', 'SST_HGF', 'Immune', 'L6_CT', 'L6_IT-1', 'L3-5_IT-3', 'Micro', 'L6_IT-2', 'Oligo', 'L5_ET', 'L3-5_IT-1', 'ADARB2_KCNG1', 'PVALB_ChC', 'OPC', 'L3-5_IT-2', 'VLMC', 'SMC', 'L6B', 'SST', 'L2-3_IT', 'LAMP5_RELN', 'PC', 'LAMP5_LHX6']


t = "L2-3_IT"

def read_sn(folder = "../Allen_MTG_Single_Cell/Subclass/Processed", skel = "DESeq2_REPLACE_Human_Chimp.txt", t = "L2-3_IT"):
    de = pd.read_csv(folder + "/" + skel.replace("REPLACE", t), sep = "\t").set_index("Gene")
    if "Allen" in folder:
        de2 = pd.read_csv(folder + "/" + skel.replace("REPLACE", t).replace("Chimp", "Gorilla"), sep = "\t").set_index("Gene")
        de5 = pd.read_csv(folder + "/" + skel.replace("REPLACE", t).replace("Chimp", "Gorilla").replace("Human", "Chimp"), sep = "\t").set_index("Gene")
    #de3 = pd.read_csv(folder + "/" + skel.replace("REPLACE", t).replace("Chimp", "Rhesus"), sep = "\t").set_index("Gene")
    #de4 = pd.read_csv(folder + "/" + skel.replace("REPLACE", t).replace("Chimp", "Marmoset"), sep = "\t").set_index("Gene")
    
    de = de[["log2FoldChange", "padj_mine"]]
    de.columns = [x + " HumChp" for x in list(de.columns)]
    if "Allen" in folder:
        de2 = de2[["log2FoldChange", "padj_mine"]]
        de2.columns = [x + " HumGor" for x in list(de2.columns)]
        de = de.join(de2)
    #de3 = de3[["log2FoldChange", "padj_mine"]]
    #de3.columns = [x + " HumRhe" for x in list(de3.columns)]
    #de4 = de4[["log2FoldChange", "padj_mine"]]
    #de4.columns = [x + " HumMarm" for x in list(de4.columns)]
    #de = de.join(de3)
    #de = de.join(de4)
    if "Allen" in folder:
        de5 = de5[["log2FoldChange", "padj_mine"]]
        de5.columns = [x + " ChpGor" for x in list(de5.columns)]
        de = de.join(de5)
    de = de.dropna()
    
    derived = []
    for index, row in de.iterrows():
        if abs(row["log2FoldChange ChpGor"]) > abs(row["log2FoldChange HumChp"]) and abs(row["log2FoldChange HumGor"]) > abs(row["log2FoldChange HumChp"]):
            derived.append("Ambiguous")
        elif abs(row["log2FoldChange ChpGor"]) < abs(row["log2FoldChange HumGor"]):
            derived.append("Human derived")
        elif abs(row["log2FoldChange ChpGor"]) > abs(row["log2FoldChange HumGor"]):
            derived.append("Chimp derived")
    de["Derived"] = derived
    return de
de = read_sn()
```

```python
def do_sign(v, lfc_col = "L2FC Chpreffed", p_col = "padj_mine Chpreffed", fdr_cut = 0.05, lfc_cut = 0, d = d_BP):
    v = v[v[p_col] < fdr_cut]
    if "Gene" not in v.columns and "gene" not in v.columns:
        v["Gene"] = v.index
    v = v[v["Gene"].isin(x)]
    upt = v[v[lfc_col] > lfc_cut].shape[0]
    downt = v[v[lfc_col] < -lfc_cut].shape[0]
    bp = upt/(upt + downt)
    bl = downt/upt
    print(bl)
    out = []
    for key in d.keys():
        if len(d[key]) > 20 and len(d[key]) < 200:
            vk = v[v["Gene"].isin(d[key])]
            if len(vk.index) > 15:
                upk = vk[vk[lfc_col] > lfc_cut]
                downk = vk[vk[lfc_col] < -lfc_cut]
                up = upk.shape[0]
                down = downk.shape[0]
                try:
                    out.append([key, up, down, binomtest(up, up + down, p = bp).pvalue, ";".join(list(upk["Gene"]) + list(downk["Gene"]))])
                except:
                    pass
    df = pd.DataFrame(out)
    if df.shape[0]:
        df = df[df[1] + df[2] > 15]
        df["l2fc"] = np.log2((df[1]/df[2]*bl))
        df.columns = ["Term", "Up", "Down", "p-value", "Genes", "l2fc"]
        return df
    else:
        return df
```
